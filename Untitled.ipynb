```python
import pandas as pd
import random
from faker import Faker
import os

# Initialize the Faker library
fake = Faker()

# List to store the data
data = []

# Function to generate a random score between 5 and 20
def random_score():
    return random.randint(7, 20)

# Function to generate a random gender
def random_gender():
    return random.choice(['male', 'female'])

# Function to generate an email with the given format
def generate_email(first_name, last_name):
    domain = random.choice(['gmail.com', 'hotmail.com'])
    email = f"{first_name.lower()}.{last_name.lower()}@{domain}"
    return email

# Function to generate a formatted telephone number
def generate_telephone():
    return fake.numerify(text="###-###-####")

# Generate 1001 rows of data
for _ in range(1001):
    id_number = fake.unique.random_number(digits=8)  # Adjusted to generate a number with up to 8 digits
    first_name = fake.first_name()
    last_name = fake.last_name()
    gender = random_gender()
    age = random.randint(19, 28)
    email = generate_email(first_name, last_name)
    street = fake.street_address()
    city = fake.city()
    state = fake.state_abbr()
    zipcode = fake.zipcode()
    telephone = generate_telephone()
    s1 = random_score()
    s2 = random_score()
    s3 = random_score()
    s4 = random_score()

    # Append the row to the data list
    data.append([id_number, first_name, last_name, gender, age, email, street, city, state, zipcode, telephone, s1, s2, s3, s4])

# Create a DataFrame
columns = ['ID(Identifiant)','FIRSTNAME', 'LASTNAME', 'gender', 'Age', 'EMAIL', 'STREET', 'CITY', 'STATE', 'ZIPCODE', 'TELEPHONE', 'S1', 'S2', 'S3', 'S4']
df = pd.DataFrame(data, columns=columns)

# Create the directory if it doesn't exist
output_dir = 'generated_data'
os.makedirs(output_dir, exist_ok=True)

# Save the DataFrame to a CSV file
output_path = os.path.join(output_dir, 'Private_student_dataset.csv')
df.to_csv(output_path, index=False)

print(f"Dataset saved to {output_path}")
```

```python
import pandas as pd
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.primitives import padding
from cryptography.hazmat.backends import default_backend
import os
```

```python
# Charger le dataset
file_path = 'Private_Students_dataset1.csv'
dataset = pd.read_csv(file_path)
# Colonnes à chiffrer
columns_to_encrypt = ["ID(Identifiant)", "Full Name", "EMAIL", "Address", "TELEPHONE"]
```

```python
key = os.urandom(32)  # Clé AES-256
iv = os.urandom(16)   # Taille du bloc AES
```

```python
# Fonction de chiffrement
def encrypt_column(column, key, iv):
    encrypted_column = []

    for value in column:
        # Convertir la valeur en bytes
        value_bytes = str(value).encode('utf-8')

        # Pad the data to be a multiple of the block size
        padder = padding.PKCS7(algorithms.AES.block_size).padder()
        padded_data = padder.update(value_bytes) + padder.finalize()

        # Créer un objet Cipher
        cipher = Cipher(algorithms.AES(key), modes.CBC(iv), backend=default_backend())
        encryptor = cipher.encryptor()
        encrypted_data = encryptor.update(padded_data) + encryptor.finalize()

        # Ajouter les données chiffrées à la colonne
        encrypted_column.append(encrypted_data.hex())

    return encrypted_column
```

```python
# Appliquer le chiffrement sur les colonnes sélectionnées
for column in columns_to_encrypt:
    if column in dataset.columns:
        dataset[column] = encrypt_column(dataset[column], key, iv)

# Sauvegarder les données chiffrées
encrypted_file_path = 'encrypted_dataset.csv'
dataset.to_csv(encrypted_file_path, index=False, sep=';')

# Sauvegarder la clé (à garder en sécurité)
with open('aes_key.bin', 'wb') as f:
    f.write(key)

# Sauvegarder le vecteur d'initialisation (à garder en sécurité)
with open('aes_iv.bin', 'wb') as f:
    f.write(iv)

print(f"Dataset chiffré sauvegardé dans {encrypted_file_path}")
print("Clé AES et IV sauvegardés dans 'aes_key.bin' et 'aes_iv.bin'")
```

```python
import pandas as pd
from cryptography.hazmat.primitives import padding
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.backends import default_backend

# Charger le dataset chiffré
encrypted_file_path = 'encrypted_dataset.csv'
dataset_encrypted = pd.read_csv(encrypted_file_path, sep=';')

# Clé et IV chargés depuis les fichiers
with open('aes_key.bin', 'rb') as f:
    key = f.read()

with open('aes_iv.bin', 'rb') as f:
    iv = f.read()

# Fonction de déchiffrement
def decrypt_column(encrypted_column, key, iv):
    decrypted_column = []

    for value_hex in encrypted_column:
        # Convertir la valeur hexadécimale en bytes
        encrypted_data = bytes.fromhex(value_hex)

        # Créer un objet Cipher
        cipher = Cipher(algorithms.AES(key), modes.CBC(iv), backend=default_backend())
        decryptor = cipher.decryptor()
        decrypted_padded_data = decryptor.update(encrypted_data) + decryptor.finalize()

        # Dépadding des données
        unpadder = padding.PKCS7(algorithms.AES.block_size).unpadder()
        decrypted_data = unpadder.update(decrypted_padded_data) + unpadder.finalize()

        # Convertir les données en str et ajouter à la colonne déchiffrée
        decrypted_column.append(decrypted_data.decode('utf-8'))

    return decrypted_column

# Colonnes à déchiffrer
columns_to_decrypt = ["ID(Identifiant)", "Full Name", "EMAIL", "Address", "TELEPHONE"]

# Appliquer le déchiffrement sur les colonnes sélectionnées
for column in columns_to_decrypt:
    if column in dataset_encrypted.columns:
        dataset_encrypted[column] = decrypt_column(dataset_encrypted[column], key, iv)

# Sauvegarder les données déchiffrées
decrypted_file_path = 'decrypted_dataset.csv'
dataset_encrypted.to_csv(decrypted_file_path, index=False)

print(f"Dataset déchiffré sauvegardé dans {decrypted_file_path}")
```

```python
import pandas as pd
import hashlib

# Chargement du jeu de données des étudiants
student_data = pd.read_csv('Private_Students_dataset1.csv')

# Étape 1 : Générer des identifiants uniques pour chaque étudiant en utilisant une fonction de hachage
def generate_pseudonym(student_id):
    hash_object = hashlib.sha256(str(student_id).encode())
    return hash_object.hexdigest()[:10]  # Utiliser les 10 premiers caractères pour le pseudonyme

student_data['ID(Identifiant)'] = student_data['ID(Identifiant)'].astype(str)  # Convertir en chaînes de caractères
student_data['Pseudo ID'] = student_data['ID(Identifiant)'].apply(generate_pseudonym)

# Étape 2 : Créer une table de correspondance sécurisée
correspondence_table = student_data[['Pseudo ID','ID(Identifiant)','Full Name', 'EMAIL', 'Address', 'TELEPHONE']]

# Stocker la table de correspondance de manière sécurisée (cryptage, accès restreint, etc.)
correspondence_table.to_csv('correspondence_table.csv', index=False)

# Réorganiser les colonnes pour placer 'Pseudo ID' au début et respecter l'ordre demandé
student_data = student_data[['Pseudo ID', 'ZIPCODE', 'gender', 'Age', 'S1', 'S2', 'S3', 'S4']]

# Sauvegarder la base de données pseudonymisée
student_data.to_csv('pseudonymized_student_data.csv', index=False)

print(f"Dataset est sauvegardé dans {'pseudonymized_student_data.csv'}")
```

```python
import pandas as pd
import numpy as np
import os

# Charger le dataset
file_path = 'dataset_apres_k_anonymate.xlsx'

# Vérifier si le fichier existe
if not os.path.isfile(file_path):
    print(f"Le fichier {file_path} est introuvable.")
else:
    try:
        # Lire le fichier XLSX
        df_k_anonymized = pd.read_excel(file_path)

        # Ajout de la colonne "Academic Level" si elle n'existe pas
        if 'Academic Level' not in df_k_anonymized.columns:
            academic_levels = ['M1', 'L1', 'L2', 'M2', 'L3']
            df_k_anonymized['Academic Level'] = np.random.choice(academic_levels, len(df_k_anonymized))

        # Vérification de la diversité
        df_grouped_diversity = df_k_anonymized.groupby(['Age', 'gender']).agg({
            'S1': 'nunique',
            'S2': 'nunique',
            'S3': 'nunique',
            'S4': 'nunique'
        }).reset_index()

        df_grouped_diversity['min_unique'] = df_grouped_diversity[['S1', 'S2', 'S3', 'S4']].min(axis=1)

        # Appliquer la condition de l-diversity (l = 3)
        l = 3
        df_l_diverse = df_grouped_diversity[df_grouped_diversity['min_unique'] >= l]

        # Merge avec l'original dataset pour obtenir les enregistrements satisfaisant l-diversity
        df_l_diverse = df_l_diverse.drop(columns=['S1', 'S2', 'S3', 'S4', 'min_unique'])
        df_l_diverse = df_k_anonymized.merge(df_l_diverse, on=['Age', 'gender'])

        # Enregistrer le DataFrame résultant dans un fichier Excel
        output_file_path = 'dataset_apres_l_diversity.xlsx'
        df_l_diverse.to_excel(output_file_path, index=False)

        print(f"Le fichier l-diversity a été enregistré à : {output_file_path}")

    except Exception as e:
        print(f"Une erreur est survenue lors de la lecture du fichier : {e}")
```

```python
import pandas as pd
import numpy as np
import os

# Charger le dataset
file_path = 'dataset_apres_k_anonymate.xlsx'

# Vérifier si le fichier existe
if not os.path.isfile(file_path):
    print(f"Le fichier {file_path} est introuvable.")
else:
    try:
        # Lire le fichier XLSX
        df_k_anonymized = pd.read_excel(file_path)

        # Fonction pour appliquer la méthode de suppression pour atteindre l-diversity
        def apply_l_diversity(df_k_anonymized, l=3):
            # Groupement par Age et Gender pour vérifier la diversité
            groups = df_k_anonymized.groupby(['Age', 'gender'])

            # Liste pour stocker les DataFrames résultants après application de l-diversity
            dfs_to_concat = []

            # Parcourir chaque groupe
            for (age, gender), group_df in groups:
                # Vérifier la diversité pour S1, S2, S3, S4
                unique_counts = group_df[['S1', 'S2', 'S3', 'S4']].nunique()

                # Vérifier si toutes les colonnes ont au moins l valeurs uniques
                if (unique_counts >= l).all():
                    # Ajouter le groupe tel quel s'il satisfait déjà l-diversity
                    dfs_to_concat.append(group_df)
                else:
                    # Tant que la diversité n'est pas atteinte, supprimer un enregistrement
                    while (unique_counts < l).any():
                        # Trouver un enregistrement à supprimer qui ne respecte pas la diversité
                        record_to_drop = group_df.iloc[0]  # Par exemple, supprimer le premier enregistrement
                        group_df = group_df.drop(record_to_drop.name)
                        unique_counts = group_df[['S1', 'S2', 'S3', 'S4']].nunique()

                    # Ajouter le groupe modifié qui satisfait l-diversity
                    dfs_to_concat.append(group_df)

            # Concaténer tous les DataFrame des groupes modifiés
            result_df = pd.concat(dfs_to_concat)

            return result_df

        # Appliquer la méthode pour atteindre l-diversity
        df_l_diverse = apply_l_diversity(df_k_anonymized, l=3)

        # Afficher le DataFrame résultant après application de la méthode de suppression pour l-diversity
        print(df_l_diverse)

    except Exception as e:
        print(f"Une erreur est survenue lors de la lecture du fichier : {e}")
```

```python
##exemple 1 ##
import pandas as pd
import numpy as np

# Charger le dataset existant
input_path = 'dataset_apres_l_diversity.xlsx'
df = pd.read_excel(input_path)

# Fonction pour appliquer la confidentialité différentielle aux codes postaux
def apply_differential_privacy(zipcode, epsilon, sensitivity, max_attempts=10):
    scale = sensitivity / epsilon
    original_length = len(str(zipcode))
    attempts = 0
    while attempts < max_attempts:
        noise = np.random.laplace(0, scale)
        new_zipcode = int(round(zipcode + noise))
        # Assurer que le code postal reste valide en fonction de sa longueur d'origine
        if original_length == 4 and 1000 <= new_zipcode <= 9999:
            return f"{new_zipcode:04d}"  # Formatage pour s'assurer de 4 chiffres
        elif original_length == 5 and 10000 <= new_zipcode <= 99999:
            return f"{new_zipcode:05d}"  # Formatage pour s'assurer de 5 chiffres
        attempts += 1
    # Si le nombre maximal de tentatives est atteint, retourner le code postal original
    return f"{zipcode:0{original_length}d}"

# Convertir la colonne ZIPCODE en entier
df['ZIPCODE'] = df['ZIPCODE'].astype(int)

# Calculer la sensibilité basée sur la plage maximale des codes postaux
sensitivity = df['ZIPCODE'].max() - df['ZIPCODE'].min()

# Appliquer la confidentialité différentielle à la colonne ZIPCODE avec epsilon = 0.1
epsilon = 0.1
df['ZIPCODE_privacy_0.1'] = df['ZIPCODE'].apply(lambda x: apply_differential_privacy(x, epsilon, sensitivity))

# Créer un DataFrame pour la comparaison des colonnes
df_comparison = df[['ZIPCODE', 'ZIPCODE_privacy_0.1']].copy()
df_comparison.columns = ['ZIPCODE_before', 'ZIPCODE_after']

# Afficher les 9 premières lignes des colonnes avant et après traitement de manière distincte
print("ZIPCODE_before traitement :")
print(df_comparison['ZIPCODE_before'].head(9))
print("\nZIPCODE_after traitement :")
print(df_comparison['ZIPCODE_after'].head(9))

# Enregistrer le DataFrame modifié dans un nouveau fichier CSV
output_path = 'Private_Students_dataset.csv'
df.to_csv(output_path, index=False)

print(f"Dataset with differential privacy applied saved to {output_path}")
```

```python
## exemple 2 ##
import pandas as pd
import numpy as np

# Charger le dataset existant
input_path = 'dataset_apres_l_diversity.xlsx'
df = pd.read_excel(input_path)

# Fonction pour calculer la sensibilité basée sur la colonne ZIPCODE
def calculate_sensitivity(df):
    return df['ZIPCODE'].max() - df['ZIPCODE'].min()

# Fonction pour appliquer la confidentialité différentielle à trois chiffres aléatoires de 0 à 9 du ZIPCODE
def apply_differential_privacy(zipcode, epsilon, sensitivity):
    # Convertir le ZIPCODE en chaîne pour manipuler chaque chiffre individuellement
    str_zipcode = str(zipcode)
    
    # Sélectionner trois positions aléatoires pour ajouter des chiffres aléatoires de 0 à 9
    num_digits = len(str_zipcode)
    positions = np.random.choice(num_digits, 3, replace=False)
    
    # Appliquer des chiffres aléatoires de 0 à 9 à chaque position sélectionnée
    new_digits = np.random.randint(0, 10, size=3)
    for i, pos in enumerate(positions):
        str_zipcode = str_zipcode[:pos] + str(new_digits[i]) + str_zipcode[pos+1:]
    
    # Reconstruire le nouveau ZIPCODE
    new_zipcode = int(str_zipcode)
    
    return new_zipcode

# Convertir la colonne ZIPCODE en entier
df['ZIPCODE'] = df['ZIPCODE'].astype(int)

# Calculer la sensibilité basée sur la plage maximale des codes postaux
sensitivity = calculate_sensitivity(df)

# Appliquer la confidentialité différentielle à trois chiffres aléatoires de 0 à 9 du ZIPCODE avec epsilon = 0.1
epsilon = 0.1
df['ZIPCODE_privacy_0.1'] = df['ZIPCODE'].apply(lambda x: apply_differential_privacy(x, epsilon, sensitivity))

# Afficher les 10 premières lignes pour vérification
print('ZIPCODE')
print(df['ZIPCODE'].head(10))
print('ZIPCODE_privacy_0.1')
print(df['ZIPCODE_privacy_0.1'].head(10))

# Enregistrer le DataFrame modifié dans un nouveau fichier CSV
output_path = 'Private_Students_dataset_modified.csv'
df.to_csv(output_path, index=False)

print(f"\nDataset with differential privacy applied saved to {output_path}")
```

```python
import pandas as pd

# Charger le dataset à partir d'un fichier CSV
input_path = 'Private_Students_dataset_final.csv'
df = pd.read_csv(input_path)

# Function to generalize the grades
def generalize_grade(grade):
    return f"[{grade},{grade+1}]"

# Apply generalization to all grade columns
for col in ['S1', 'S2', 'S3', 'S4']:
    df[col] = df[col].apply(generalize_grade)

# Export to a CSV file
df.to_csv("generalized_data.csv", index=False)

# Import the data again to ensure it's correctly exported
imported_df = pd.read_csv("generalized_data.csv")

print(imported_df)

# Enregistrer le DataFrame modifié dans un fichier CSV
output_path = 'Private_Students_dataset_modified_final.csv'
df.to_csv(output_path, index=False)

print(f"Dataset généralisé enregistré dans {output_path}")
```
